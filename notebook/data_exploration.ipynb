```python
import pandas as pd
import os

RAW_DIR = "../data/raw"

# UN/LOCODE files
locode_codes = pd.read_csv(f"{RAW_DIR}/UN_LOCODE/code-list.csv")
locode_countries = pd.read_csv(f"{RAW_DIR}/UN_LOCODE/country-codes (1).csv")

# WPI
wpi = pd.read_csv(f"{RAW_DIR}/WPI/world_port_index.csv")

print("LOCODE code-list shape:", locode_codes.shape)
print("LOCODE countries shape:", locode_countries.shape)
print("WPI shape:", wpi.shape)
```

```python
print("=== LOCODE code-list columns ===")
print(locode_codes.columns.tolist())
print(locode_codes.head(3))

print("\n=== WPI columns ===")
print(wpi.columns.tolist())
print(wpi.head(3))
```

```python
# We need to find common keys to merge on — likely port name / country code / UN/LOCODE
print(locode_codes.dtypes)
print(wpi.dtypes)
```

```python
print(locode_codes.isnull().sum())
print(wpi.isnull().sum())
```

```python
print(wpi['UN/LOCODE'].head(10))
print("Nulls in WPI UN/LOCODE:", wpi['UN/LOCODE'].isnull().sum())
print("Nulls in WPI Latitude/Longitude:", wpi[['Latitude','Longitude']].isnull().sum())
```

```python
locode_codes['UNLOCODE'] = locode_codes['Country'].astype(str) + locode_codes['Location'].astype(str)
print(locode_codes['UNLOCODE'].head(10))
print("Nulls:", locode_codes['UNLOCODE'].isnull().sum())
```
